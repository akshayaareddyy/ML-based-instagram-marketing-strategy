## Web Scraping Data Collection - Yahoo Finance

```python
import numpy as np
import requests
import pandas as pd
np.random.seed(42)
```

```python
# Data URL
url = "https://query1.finance.yahoo.com/v1/finance/screener/predefined/saved"

# Web Scrapping Parameters
params = {
    "scrIds": "most_expensive_stocks",
    "count": 100,
    "start": 0
}

headers = {
    "User-Agent": "Mozilla/5.0"
}
```

```python
# Data Extraction from URL
all_rows = []
start = 0
count = 100

# Looping over stocks
while True:
    params["start"] = start
    response = requests.get(url, params=params, headers=headers).json()
    quotes = response["finance"]["result"][0]["quotes"]

    if not quotes:
        break

    all_rows.extend(quotes)
    start += count

```

## Web Scraping Data Collection through yFinance Python Package

```python
import yfinance as yf
import pandas as pd
from tqdm import tqdm

# Converting symbols to list and taking first 1300 due to rate limit
symbols = data_yf.symbol.tolist()[0:1299]

# Setting up storage
results = []

# Loop over symbols
for sym in tqdm(symbols, desc="sectors"):
    try:
        ticker = yf.Ticker(sym)
        info = ticker.info

        sector = info.get("sector")
        industry = info.get("industry")

        results.append({
            "symbol": sym,
            "sector": sector,
            "industry": industry
        })
    except Exception as e:
        print(f"Error fetching {sym}: {e}")
        results.append({
            "symbol": sym,
            "sector": None,
            "industry": None
        })

# Converting results to DataFrame
sector_industry_yf = pd.DataFrame(results)

```

## API Data Access - Instagram

### Data Extraction Function

```python
import requests
import pandas as pd
import time

def fetch_hashtag(hashtag, access_token, user_id, limit=50, sleep_time=1):

    print(f"Fetching data for #{hashtag}")

    # Getting hashtag ID
    search_url = "https://graph.facebook.com/v18.0/ig_hashtag_search"
    params = {
        "user_id": user_id,
        "q": hashtag,
        "access_token": access_token
    }

    response = requests.get(search_url, params=params).json()

    if "data" not in response or len(response["data"]) == 0:
        raise Exception(f"Error fetching hashtag ID: {response}")

    hashtag_id = response["data"][0]["id"]
    print(f"Hashtag ID: {hashtag_id}")

    # Getting Data
    media_url = f"https://graph.facebook.com/v18.0/{hashtag_id}/recent_media"
    params = {
        "user_id": user_id,
        "fields": "id,media_type,timestamp,like_count,comments_count",
        "access_token": access_token,
        "limit": limit
    }

    all_posts = []

    while True:
        response = requests.get(media_url, params=params).json()

        if "error" in response:
            print("Error:", response["error"])
            break

        data = response.get("data", [])

        for post in data:
            post["hashtag"] = f"#{hashtag}"

        all_posts.extend(data)

        print(f"Collected: {len(all_posts)} posts")

        # Pagination
        paging = response.get("paging", {})
        next_page = paging.get("next")

        if not next_page:
            print("No more pages.")
            break

        media_url = next_page
        params = None

        time.sleep(sleep_time)

    return pd.DataFrame(all_posts)
```
